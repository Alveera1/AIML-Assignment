```python
import cv2
from ultralytics import YOLO
```

```python
model = YOLO("yolo11n.pt")
```

```python
video = cv2.VideoCapture("cctv.mp4")

if not video.isOpened():
    print("Error: Could not open CCTV footage.")
else:
    print("CCTV footage opened successfully!")
```

```python
while True:
    ret, frame = video.read()

    if not ret:
        break

    results = model(frame)

    detected_frame = results[0].plot()

    cv2.imshow("CCTV Object Detection", detected_frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

video.release()
cv2.destroyAllWindows()
```
